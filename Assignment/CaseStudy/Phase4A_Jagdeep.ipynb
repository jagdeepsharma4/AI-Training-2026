PII Masking using Python Regex

```python
import re
import pandas as pd
```

```python
def mask_pii(text):

    if not isinstance(text, str):
        return ""

    # Mask Aadhaar numbers (XXXX XXXX XXXX)
    text = re.sub(
        r"\b[2-9]\d{3}\s?\d{4}\s?\d{4}\b",
        "[AADHAAR_MASKED XXXX]",
        text
    )

    # Mask Indian mobile numbers
    text = re.sub(
        r"(?<!\d)(?:\+91[-\s]?|0)?[6-9]\d{9}(?!\d)",
        "[PHONE_MASKED XXXX]",
        text
    )

    # Mask email addresses
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        "[EMAIL_MASKED XXXX]",
        text
    )

    # Mask explicitly labeled home addresses
    text = re.sub(
        r"(?i)\b(?:home address|residential address|my address)\s*:"
        r"\s*[^\n.;]+",
        "[ADDRESS_MASKED XXXX]",
        text
    )

    return text
```

```python
complaint = """
My name is Ravi Kumar.

My Aadhaar number is 2345 6789 1234.
My mobile number is 9876543210.
My email is ravi@example.com.

My address: House No 25, Main Road, Shimla.

I have not received electricity for four days.
"""

masked_complaint = mask_pii(complaint)

print("Original Complaint:\n")
print(complaint)

print("\nMasked Complaint:\n")
print(masked_complaint)
```

```python
import pandas as pd

# Load previously saved DataFrame
df = pd.read_csv("jan_samadhan_cleaned.csv")

# Apply PII masking
df["Masked_Complaint"] = (
    df["Complaint_Description"]
    .fillna("")
    .apply(mask_pii)
)

# Display results
display(df[["Complaint_Description", "Masked_Complaint"]].head())
```
